# Rule based approach to find call to order in text:

```python
import re
```

```python
def contains_cto(dct):
    rg1 = r'^(?!.*\b(?:keinen|kein|erteilten)\b)(?=.*\berteil(?:e|en)\b)(?!.*\bnicht\b).*\bordnungsruf\b'
    rg2 = r'.*rufe.*\bzur\sordnung\b.*'
    text = dct.get('text', '')
    return bool(re.search(rg1, text, re.IGNORECASE) or re.search(rg2, text, re.IGNORECASE))
```

```python
example_dict = data.speech.iloc[0]
for dct in example_dict:
    print(dct, contains_cto(dct))
```
